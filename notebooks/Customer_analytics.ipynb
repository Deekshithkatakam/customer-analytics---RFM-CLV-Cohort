RFM Feature engineering
it gives an idea on which customers matter most to the business and how should each customer treated diffrently

Metrics
Recency - days since last purchase( it tells about customer is still active or not)
Frequency - number of distinct orders (diffrentiates loya repeat,buyer vs one_time purchase)
monetary - total revenue (how much do they actually bring in)
tenure - days since first purchase(new vs long term)
AOV - average_order_value - they buy a little or lot per order
return rate - cancelled/total orders are they clean customer or one who buys and sends a lot back.



Recency: since this dataset ends in Dec 2011 rather than today, Recency is calculated relative to the last transaction date in the data (+1 day)

```python
reference_date = Sales["InvoiceDate"].max() + pd.Timedelta(days =1)
```

```python
RFM = Sales.groupby("Customer_ID").agg(
    Recency = ("InvoiceDate", lambda x: (reference_date - x.max()).days),
    Frequency = ("Invoice","nunique"),
    Monetary = ("Revenue", "sum"),
    First_purchase = ("InvoiceDate","min"),
    Last_purchase = ("InvoiceDate","max"),
    Total_items = ("Quantity","sum")
    ).reset_index()
```

```python
RFM.head()
```

Tenure — days since first purchase → new customer vs. long-term?
AOV — average order value → do they buy a little or a lot per order?

```python
RFM["Tenure"] = (reference_date - RFM["First_purchase"]).dt.days
RFM["AOV"] = RFM["Monetary"] / RFM["Frequency"]
```

Return rate  cancellations weere kept seperately from sales 
Here we count each customers no.of of unique invoices from cancellations and later merge that with RFM Table and compute the return Rate 

RETURN_RATE = CANCELLED_ORDERS /  (frequency +  cancelled_orders)
 first we have to create cancell counts and then merge it with RFM

```python
Cancel_counts = Cancellation.groupby("Customer_ID")["Invoice"].nunique().reset_index()
```

```python
Cancel_counts.columns = ["Customer_ID","Cancelled_orders"]
```

```python
RFM = RFM.merge(Cancel_counts, on = "Customer_ID", how = "left")
```

```python
RFM.info() 
```

```python
RFM["Cancelled_orders"] = RFM["Cancelled_orders"].fillna(0)
```

```python
RFM["Return_Rate"] = RFM["Cancelled_orders"] / (RFM["Frequency"] + RFM["Cancelled_orders"])
```

```python
RFM.head()
```

```python
print("RFM shape:", RFM.shape)
RFM[["Recency","Frequency","Monetary","Tenure","AOV","Return_Rate"]].describe()
```

```python
RFM.to_csv("../Data/processed/customer_features.csv",index = False)
print("Saved:", RFM.shape)
```

RFM Scoring
Ranks every customer into quintilies (1-5) on Recency, Frequency, Monetary then later on three scores we will segment customers with labels eg. champions at risk, lost numbers into groups marketimng can actually act on 
target high-value segments win back at risk ones stop spending on lost customers

```python
RFM["R_Score"] = pd.qcut(RFM["Recency"], 5 , labels = [5,4,3,2,1]).astype(int)
```

```python
RFM["F_Score"] = pd.qcut(RFM["Frequency"].rank(method = "first"),5, labels = [1,2,3,4,5]).astype(int)
```

```python
RFM["M_Score"] = pd.qcut(RFM["Monetary"], 5 , labels = [1,2,3,4,5]).astype(int)
```

labelling the segments 
RFM Segment label
1. r>=4 , f>=4 , m>= 4 --- champions
2. r>=3 , f>=3 , m>= 3 --- Loyal customers
3. r>=4 , f<=2 , --- New_Customers
4. r<=2, f>=3, ---- at risk
5. r<=2, f<=2 and m<=2 ---- lost
6. anthing else needs attention

```python
def segment_customer(x):
    r,f,m = x["R_Score"], x["F_Score"], x["M_Score"]
    if r>=4 and f>=4 and m>=4:
        return "Champions"
    elif r>=3 and f>=3 and m>=3:
        return "Loyal_customers"
    elif r>=4 and f<=2:
        return "New_customers"
    elif r<=2 and f>=3:
        return "At_Risk"
    elif r<=2 and f<=2 and m<=2:
        return "lost" 
    else:
        return "Needs_Attention"
```

```python
RFM["Segment"] = RFM.apply(segment_customer,axis =1)
```

```python
Segment_counts = RFM["Segment"].value_counts()
```

```python
segment_pct = (Segment_counts/len(RFM)*100).round(1)
segment_pct
```

```python
Segment_summary = Segment_counts.reset_index()
Segment_summary.columns = ["Segment","Customer_count"]
Segment_summary["Pct_customers"] = segment_pct.values
Segment_summary
```

Cohort / Retention Analysis

Once a customer joins how long do they actually keep buying, this tells are we retaining and customers overtime or losing most of them quickly after thier first purchase

RFM and CLV looked after customer and segmentation but here 
cohort groups customers by when they joined a cohort means how many customers joined or made thier first purchase in same month
then we will look after the percentage of customers are still active after first purchase

if say only 20% of customers are still active 3 months after thier first purchase thats a retention problem woth solving

process- 
cohort retention table - rows are month a customer first purchased
columns - months since that first purchase
values - % of that cohort still active

```python
Sales["Order_Month"] = Sales["InvoiceDate"].dt.to_period("M")
```

```python
cohort_map = Sales.groupby("Customer_ID")["Order_Month"].min().reset_index()
cohort_map.columns = ["Customer_ID","Cohort_Month"]
```

This is a separate, smaller table — one row per customer, holding just their Cohort_Month (the month of their very first purchase, computed by finding the minimum Order_Month across all their transactions).

```python
## Lets merge sales with Sales so that we can have clear idea on To answer "how many months after joining did this transaction happen," you need both pieces of information on the same row:
Sales = Sales.merge(cohort_map, on="Customer_ID", how="left")
```

```python
Sales[["Customer_ID","Cohort_Month","Order_Month"]].sort_values("Customer_ID", ascending = True)
```

```python
Sales 
```

```python
Sales["Cohort_Index"] = ((Sales["Order_Month"].dt.year - Sales["Cohort_Month"].dt.year)*12 + (Sales["Order_Month"].dt.month - Sales["Cohort_Month"].dt.month))
```

```python
Sales[["Customer_ID","Cohort_Month","Order_Month","Cohort_Index"]].head()
```

Lets build a retention table

```python
cohort_data = Sales.groupby(["Cohort_Month","Cohort_Index"])["Customer_ID"].nunique().reset_index()
cohort_pivot = cohort_data.pivot(index="Cohort_Month",  columns = "Cohort_Index",values = "Customer_ID")
```

```python
cohort_pivot
```

```python
cohort_size = cohort_pivot.iloc[:,0]
```

```python
retention_table  = cohort_pivot.divide(cohort_size,axis  = 0).round(3)*100
```
